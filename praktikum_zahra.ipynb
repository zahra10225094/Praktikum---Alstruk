```python
# Praktikum 1: Mengenali Tipe Data

nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

print(nama)
print(umur)
print(berat)
print(terdaftar)

print(type(nama))
print(type(umur))
print(type(berat))
print(type(terdaftar))

# mengubah umur menjadi 20
umur = 20
print ("umur setelah diubah:", umur)
```

```python
# Praktikum 2: Operator Aritmatika

harga = 12500
jumlah = 4

total = harga * jumlah
sisa = total % 2000
harga_diskon = total - 5000

print("Harga per buku: Rp", harga)
print("Jumlah buku:", jumlah)
print("Total harga: Rp", total)
print("Sisa pembagian dengan 2000:", sisa)
print("Harga setelah diskon: Rp", harga_diskon)
```

```python
# Praktikum 3: Operator Perbandingan dan Logika

nilai = 82
kehadiran = 90

syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

lulus = syarat_nilai and syarat_kehadiran

print("Nilai:", nilai)
print("Kehadiran:", kehadiran, "%")
print("Memenuhi syarat nilai:", syarat_nilai)
print("Memenuhi syarat kehadiran:", syarat_kehadiran)
print("Lulus praktikum:", lulus)

# Mengubah nilai menjadi 70
nilai = 70

syarat_nilai = nilai >= 75
lulus_and = syarat_nilai and syarat_kehadiran
lulus_or = syarat_nilai or syarat_kehadiran

print("\nSetelah nilai diubah menjadi 70:")
print("Nilai:", nilai)
print("Memenuhi syarat nilai:", syarat_nilai)
print("Lulus dengan and:", lulus_and)
print("Lulus dengan or:", lulus_or)
```

```python
# Praktikum 4: Operator Penugasan

poin = 100
print("Nilai awal:", poin)

poin += 25
print("Setelah poin += 25:", poin)

poin -= 10
print("Setelah poin -= 10:", poin)

poin *= 2
print("Setelah poin *= 2:", poin)

poin /= 2
print("Setelah poin /= 2:", poin)
```

```python
# Praktikum 5: Struktur Data Dasar

nilai = [78, 85, 90, 85]
mahasiswa = {
    "nama": "Nadia",
    "kelas": "IF-A"
}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

# Menampilkan nilai pertama dari list
print("Nilai pertama:", nilai[0])

# Menambahkan nilai 95
nilai.append(95)
print("List terbaru:", nilai)

# Menampilkan nama mahasiswa
print("Nama mahasiswa:", mahasiswa["nama"])

# Menampilkan koordinat pertama
print("Koordinat pertama:", koordinat[0])

# Menampilkan angka unik secara berurutan
print("Angka unik:", sorted(angka))
```

```python
# Tantangan Terintegrasi: Program Rekap Nilai

nama = "Zahra"

# Pengujian pertama
nilai_tugas = 80
nilai_ujian = 90

rata_rata = (nilai_tugas + nilai_ujian) / 2
lulus = rata_rata >= 75

print("=== Rekap Nilai Mahasiswa ===")
print("Nama:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)

# Pengujian kedua
nilai_tugas = 60
nilai_ujian = 50

rata_rata = (nilai_tugas + nilai_ujian) / 2
lulus = rata_rata >= 75

print("\n=== Pengujian Kedua ===")
print("Nama:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)
```
